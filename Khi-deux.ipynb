```python
import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency

# Données de l'exemple sous forme de tableau (observé)
# Lignes : Tranches d'âge (< 35 ans, ≥ 35 ans)
# Colonnes : Types de logement (Appartement, Maison, Loft)

data = np.array([
    [20, 10, 10],  # < 35 ans
    [10, 25, 5]    # ≥ 35 ans
])

# Création d'une DataFrame pour l'affichage
df = pd.DataFrame(data, 
                  index=["< 35 ans", "≥ 35 ans"], 
                  columns=["Appartement", "Maison", "Loft"])

print("Tableau observé :")
print(df)

# Test du Khi² d'indépendance
chi2, p_value, dof, expected = chi2_contingency(data)

# Affichage des résultats
print("\nRésultats du test du Khi² :")
print(f"Statistique Khi² : {chi2:.2f}")
print(f"p-value : {p_value:.4f}")
print(f"Degrés de liberté : {dof}")
print("\nTableau des effectifs attendus (sous H0) :")
print(pd.DataFrame(expected, index=["< 35 ans", "≥ 35 ans"], columns=["Appartement", "Maison", "Loft"]))

# Interprétation
alpha = 0.05
if p_value < alpha:
    print("\nConclusion : Il y a une dépendance entre la tranche d'âge et le type de logement.")
else:
    print("\nConclusion : Aucune dépendance significative détectée entre l'âge et le type de logement.")
```
