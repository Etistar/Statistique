```python
import numpy as np
import statsmodels.stats.api as sms

# Calcul avec la librairie statsmodels

# Exemple de données : temps moyen passé sur TikTok par jour (en minutes) d’un échantillon de 20 ados
tiktok_temps = [95, 100, 92, 110, 88, 97, 102, 99, 105, 96, 93, 98, 91, 90, 108, 101, 94, 89, 99, 103]

# Calcul de l'intervalle de confiance à 95 %
intervalle = sms.DescrStatsW(tiktok_temps).tconfint_mean()

print(f"Intervalle de confiance à 95 % : [{intervalle[0]:.2f}, {intervalle[1]:.2f}] minutes")
```

```python
import numpy as np
import scipy.stats as stats

# Taille de l'échantillon
n = len(tiktok_temps)

# Moyenne et écart-type de l'échantillon
moyenne = np.mean(tiktok_temps)
ecart_type = np.std(tiktok_temps, ddof=1)  # ddof=1 pour l'écart-type de l'échantillon

# Niveau de confiance
alpha = 0.05  # pour un intervalle à 95 %

# Valeur critique de la loi de Student
t_critique = stats.t.ppf(1 - alpha/2, df=n-1)

# Marge d'erreur
marge = t_critique * (ecart_type / np.sqrt(n))

# Intervalle de confiance
borne_inf = moyenne - marge
borne_sup = moyenne + marge

print(f"Intervalle de confiance à 95 % : [{borne_inf:.2f}, {borne_sup:.2f}]")
```
