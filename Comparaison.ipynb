```python
import numpy as np
import math
from scipy.stats import t

# Données
m1 = 425000     # moyenne groupe 1 (près du métro)
m2 = 405000     # moyenne groupe 2 (loin du métro)
s1 = 25000      # écart-type groupe 1
s2 = 30000      # écart-type groupe 2
n1 = n2 = 40    # tailles d'échantillon égales

# Étape 1 : calcul de la variance combinée (pooled)
sp_squared = (((n1 - 1)*s1**2 + (n2 - 1)*s2**2) / (n1 + n2 - 2))
sp = math.sqrt(sp_squared)

# Étape 2 : calcul du t-statistique
t_stat = (m1 - m2) / (sp * math.sqrt(1/n1 + 1/n2))

# Étape 3 : calcul du seuil critique ou p-value
ddl = n1 + n2 - 2
p_value = 2 * (1 - t.cdf(abs(t_stat), df=ddl))

print(f"Statistique t : {t_stat:.2f}")
print(f"p-value : {p_value:.4f}")
```

```python
from scipy.stats import ttest_ind

# Génération de données simulées (même moyennes et écarts-types que plus haut)
groupe_metro = np.random.normal(425000, 25000, 40)
groupe_loin = np.random.normal(405000, 30000, 40)

# Test t de Student pour échantillons indépendants
t_stat, p_value = ttest_ind(groupe_metro, groupe_loin, equal_var=True)

print(f"Statistique t : {t_stat:.2f}")
print(f"p-value : {p_value:.4f}")
```

```python
p-value < 0.05 → la différence est statistiquement significative
```
